#Step 1 - Import required libraries

```python
# ============================================================
# Step 1: Import required libraries
# ============================================================


# Import the Google Drive
from google.colab import drive

# Import Pandas for loading, checking and combining CSV datasets
import pandas as pd


# Import OS for working with file paths and folders
import os
```

#Step 2 - Connect Google Drive

```python
# ============================================================
# Step 2: Connect Google Drive
# ============================================================


# Connect Google Drive to the Colab notebook
drive.mount(
    "/content/drive"
)


# Confirm that the connection step is complete
print(
    "Google Drive connected"
)
```

#Step 3 - Set the Simple and Semantic CSV locations

```python
# ============================================================
# Step 3: Set the Simple and Semantic CSV locations
# ============================================================


# Set the location of the Simple Features CSV
simple_csv_path = (
    "/content/drive/MyDrive/DISSERTATION/DATASET/RAW DATASET/Web robot detection - Server logs/simple_features.csv"
)


# Set the location of the Semantic Features CSV
semantic_csv_path = (
    "/content/drive/MyDrive/DISSERTATION/DATASET/RAW DATASET/Web robot detection - Server logs/semantic_features.csv"
)


# Check whether the Simple Features CSV exists
print(
    "Simple CSV found:",
    os.path.exists(simple_csv_path)
)


# Check whether the Semantic Features CSV exists
print(
    "Semantic CSV found:",
    os.path.exists(semantic_csv_path)
)
```

#Step 4 - Load both CSV files

```python
# ============================================================
# Step 4: Load both CSV files
# ============================================================


# Load the Simple Features CSV into a Pandas DataFrame
simple_dataframe = pd.read_csv(
    simple_csv_path
)


# Load the Semantic Features CSV into a Pandas DataFrame
semantic_dataframe = pd.read_csv(
    semantic_csv_path
)


# Confirm that both files were loaded
print(
    "Simple and Semantic CSV files loaded successfully"
)
```

#Step 5 - Check number of rows and columns

```python
# ============================================================
# Step 5: Check number of rows and columns
# ============================================================


# Display the number of rows and columns in the Simple Features dataset
print(
    "Simple Features size:",
    simple_dataframe.shape
)


# Display the number of rows and columns in the Semantic Features dataset
print(
    "Semantic Features size:",
    semantic_dataframe.shape
)
```

#Step 6 - Inspect column names

```python
# ============================================================
# Step 6: Inspect column names
# ============================================================


# Display all column names in the Simple Features dataset
print(
    "Simple Features columns:"
)

print(
    simple_dataframe.columns.tolist()
)


# Display all column names in the Semantic Features dataset
print(
    "\nSemantic Features columns:"
)

print(
    semantic_dataframe.columns.tolist()
)
```

#Step 7 - Check the ID and label columns

```python
# ============================================================
# Step 7: Check the ID and label columns
# ============================================================


# Check whether ID exists in the Simple Features dataset
print(
    "ID in Simple:",
    "ID" in simple_dataframe.columns
)


# Check whether ID exists in the Semantic Features dataset
print(
    "ID in Semantic:",
    "ID" in semantic_dataframe.columns
)


# Check whether ROBOT exists in the Simple Features dataset
print(
    "ROBOT in Simple:",
    "ROBOT" in simple_dataframe.columns
)


# Check whether ROBOT exists in the Semantic Features dataset
print(
    "ROBOT in Semantic:",
    "ROBOT" in semantic_dataframe.columns
)
```

#Step 8 - Check duplicate session IDs

```python
# ============================================================
# Step 8: Check duplicate session IDs
# ============================================================


# Select the ID column from the Simple Features dataset
simple_duplicate_ids = (

    simple_dataframe["ID"]

    # Mark IDs that appear more than once
    .duplicated()

    # Count the total number of duplicated IDs
    .sum()
)


# Select the ID column from the Semantic Features dataset
semantic_duplicate_ids = (

    semantic_dataframe["ID"]

    # Mark IDs that appear more than once
    .duplicated()

    # Count the total number of duplicated IDs
    .sum()
)


# Display the number of duplicate IDs in the Simple Features dataset
print(
    "Duplicate IDs in Simple:",
    simple_duplicate_ids
)


# Display the number of duplicate IDs in the Semantic Features dataset
print(
    "Duplicate IDs in Semantic:",
    semantic_duplicate_ids
)
```

#Step 9 - Verify whether both files contain the same session IDs

```python
# ============================================================
# Step 9: Verify the same sessions exist in both files
# ============================================================


# Convert the Simple session IDs into a set
simple_ids = set(
    simple_dataframe["ID"]
)


# Convert the Semantic session IDs into a set
semantic_ids = set(
    semantic_dataframe["ID"]
)


# Check whether both datasets contain exactly the same IDs
same_session_ids = (
    simple_ids == semantic_ids
)


# Display the result
print(
    "Both files contain the same session IDs:",
    same_session_ids
)
```

#Step 10 - Check for unmatched sessions

```python
# ============================================================
# Step 10: Check for unmatched sessions
# ============================================================


# Find session IDs that exist in Simple but not in Semantic
simple_only_ids = (

    # Subtract the Semantic ID set from the Simple ID set
    simple_ids - semantic_ids
)


# Find session IDs that exist in Semantic but not in Simple
semantic_only_ids = (

    # Subtract the Simple ID set from the Semantic ID set
    semantic_ids - simple_ids
)


# Display the number of sessions found only in Simple
print(
    "Sessions only in Simple:",

    # Count the number of unmatched Simple session IDs
    len(simple_only_ids)
)


# Display the number of sessions found only in Semantic
print(
    "Sessions only in Semantic:",

    # Count the number of unmatched Semantic session IDs
    len(semantic_only_ids)
)
```

#Step 11 - Verify that the Human or Bot labels match

```python
# ============================================================
# Step 11: Verify that the class labels match
# ============================================================


# Select only the ID and ROBOT columns from the Simple dataset
simple_labels = simple_dataframe[

    # Keep the session ID and class label columns
    ["ID", "ROBOT"]

# Create a separate copy of the selected columns
].copy()


# Select only the ID and ROBOT columns from the Semantic dataset
semantic_labels = semantic_dataframe[

    # Keep the session ID and class label columns
    ["ID", "ROBOT"]

# Create a separate copy of the selected columns
].copy()


# Join the Simple and Semantic label tables using the session ID
label_check = pd.merge(

    # Use the Simple ID and label table
    simple_labels,

    # Use the Semantic ID and label table
    semantic_labels,

    # Match the rows using the ID column
    on="ID",

    # Keep IDs that exist in both tables
    how="inner",

    # Rename the two ROBOT columns so they can be identified separately
    suffixes=(
        "_simple",
        "_semantic"
    )
)


# Compare the Simple and Semantic labels for every matched session
label_mismatches = (

    # Select the ROBOT label from the Simple dataset
    label_check["ROBOT_simple"]

    # Check whether it is different from the Semantic label
    !=

    # Select the ROBOT label from the Semantic dataset
    label_check["ROBOT_semantic"]

# Count the total number of label mismatches
).sum()


# Display the number of sessions with different labels
print(

    # Display a description of the result
    "Label mismatches:",

    # Display the total number of mismatched labels
    label_mismatches
)
```

#Step 12 - Check columns shared by both datasets

```python
# ============================================================
# Step 12: Check columns shared by both datasets
# ============================================================


# Get all column names from the Simple Features dataset
common_columns = set(
    simple_dataframe.columns

# Find the column names that also exist in the Semantic Features dataset
).intersection(
    semantic_dataframe.columns
)


# Display a heading for the shared columns
print(
    "Columns present in both files:"
)


# Sort and display the shared column names
print(
    sorted(common_columns)
)
```

#Step 13 - Remove duplicated ROBOT label before combining

```python
# ==================================================================================
# Step 13: Keep ID for matching and remove duplicate ROBOT label before combining
# ==================================================================================


# Remove ROBOT from Semantic because the same label already exists in Simple
semantic_features_only = semantic_dataframe.drop(

    # Remove only the repeated ROBOT column
    columns=[
        "ROBOT"
    ]
)


# Display the Semantic columns that will be used for combining
print(
    "Semantic columns ready for combining:"
)


# Show the remaining Semantic column names
print(
    semantic_features_only.columns.tolist()
)
```

#Step 14 - Combine Simple and Semantic datasets

```python
# ============================================================
# Step 14: Combine Simple and Semantic by ID
# ============================================================


# Combine the Simple and Semantic datasets
simple_semantic_dataframe = pd.merge(

    # Use Simple as the main dataset
    simple_dataframe,

    # Add the Semantic features without the repeated ROBOT column
    semantic_features_only,

    # Match sessions using their ID
    on="ID",

    # Keep sessions that are present in both datasets
    how="inner",

    # Make sure each ID matches only one ID
    validate="one_to_one"
)


# Display the number of rows and columns after combining
print(
    "Combined dataset size:",
    simple_semantic_dataframe.shape
)


# Display the first Ten rows of the combined dataset
display(
    simple_semantic_dataframe.head(10)
)
```

#Step 15 - Check the combined dataset

```python
# ============================================================
# Step 15: Check the combined dataset
# ============================================================


# Display the number of rows and columns in the combined dataset
print(

    # Display a description for the result
    "Combined dataset size:",

    # Get the number of rows and columns
    simple_semantic_dataframe.shape
)


# Display the number of unique session IDs
print(

    # Display a description for the result
    "Unique session IDs:",

    # Select the ID column and count unique IDs
    simple_semantic_dataframe["ID"].nunique()
)


# Display the number of duplicate session IDs
print(

    # Display a description for the result
    "Duplicate session IDs:",

    # Select the ID column, find duplicates and count them
    simple_semantic_dataframe["ID"].duplicated().sum()
)


# Display a heading for the Human and Bot class counts
print(
    "\nClass label counts:"
)


# Select the ROBOT column and count each class label
print(
    simple_semantic_dataframe["ROBOT"].value_counts()
)


# Display a heading for the first five rows
print(
    "\nFirst five rows:"
)


# Select and display the first five rows of the combined dataset
display(
    simple_semantic_dataframe.head()
)
```

#Step 16 - Verify all sessions were combined

```python
# ============================================================
# Step 16: Verify all sessions were combined
# ============================================================


# Create a variable to store the number of Simple sessions
simple_session_count = len(

    # Count all rows in the Simple dataset
    simple_dataframe
)


# Create a variable to store the number of combined sessions
combined_session_count = len(

    # Count all rows in the combined dataset
    simple_semantic_dataframe
)


# Create a variable to store the number of sessions lost
sessions_lost = (

    # Take the original number of Simple sessions
    simple_session_count

    # Subtract the number of combined sessions
    -
    combined_session_count
)


# Display the original Simple session count
print(

    # Display the result name
    "Original Simple sessions:",

    # Display the number of Simple sessions
    simple_session_count
)


# Display the combined session count
print(

    # Display the result name
    "Combined sessions:",

    # Display the number of combined sessions
    combined_session_count
)


# Display how many sessions were lost
print(

    # Display the result name
    "Sessions lost during combining:",

    # Display the calculated number of lost sessions
    sessions_lost
)
```

#Step 17 - Save the combined Simple + Semantic as a new CSV

```python
# ============================================================
# Step 17: Save the combined dataset
# ============================================================


# Create the file path where the combined CSV will be saved
combined_csv_path = (

    # Set the Google Drive location and CSV file name
    "/content/drive/MyDrive/DISSERTATION/DATASET/PROCESSED CSV FILES/simple_semantic_combined.csv"
)


# Save the combined DataFrame as a CSV file
simple_semantic_dataframe.to_csv(

    # Save the CSV in the selected location
    combined_csv_path,

    # Do not include Pandas row numbers in the CSV
    index=False
)


# Display a message to confirm that the file was saved
print(
    "Combined CSV saved successfully"
)


# Display the location of the saved CSV
print(

    # Display the result name
    "Saved location:",

    # Display the file path
    combined_csv_path
)


# Display the number of rows and columns in the saved dataset
print(

    # Display the result name
    "Final CSV size:",

    # Display the rows and columns
    simple_semantic_dataframe.shape
)
```
