#STEP 8 - Create a function to calculate mouse features

```python
# ============================================================
# STEP 8 : FUNCTION TO CALCULATE MOUSE FEATURES
# ============================================================


# Create a pattern to identify mouse movement coordinates
# stored in the format [m(x,y)]
mouse_coordinate_pattern = re.compile(
    r'\[m\(\s*(-?\d+)\s*,\s*(-?\d+)\s*\)\]'
)


# Create a function that receives mouse behaviour
# and mouse timestamp data
def calculate_mouse_features(
    mouse_behaviour,
    mouse_times
):


    # --------------------------------------------------------
    # 1. EXTRACT MOUSE COORDINATES
    # --------------------------------------------------------

    # Convert missing mouse behaviour into empty text
    if mouse_behaviour is None:
        mouse_behaviour = ""

    # Make sure mouse behaviour is text
    mouse_behaviour = str(
        mouse_behaviour
    )


    # Extract all [m(x,y)] coordinates
    mouse_coordinates = [

        (
            int(x),
            int(y)
        )

        for x, y in mouse_coordinate_pattern.findall(
            mouse_behaviour
        )
    ]


    # --------------------------------------------------------
    # 2. COUNT MOUSE MOVEMENT EVENTS
    # --------------------------------------------------------

    # Number of mouse movement coordinates
    mouse_events = len(
        mouse_coordinates
    )


    # --------------------------------------------------------
    # 3. CALCULATE TOTAL MOUSE DISTANCE
    # --------------------------------------------------------

    # Start total distance at zero
    total_distance = 0.0


    # Compare every mouse position with the next position
    for first_point, second_point in zip(
        mouse_coordinates,
        mouse_coordinates[1:]
    ):


        # Separate first coordinate
        first_x, first_y = first_point


        # Separate second coordinate
        second_x, second_y = second_point


        # Calculate Euclidean distance
        distance_between_points = math.sqrt(

            (second_x - first_x) ** 2

            +

            (second_y - first_y) ** 2
        )


        # Add the distance to the session total
        total_distance += distance_between_points


    # --------------------------------------------------------
    # 4. READ AND CLEAN MOUSE TIMESTAMPS
    # --------------------------------------------------------

    # Create empty list for valid timestamps
    mouse_timestamps = []


    # Convert missing timestamp data into empty text
    if mouse_times is None:
        mouse_times = ""


    # Make sure timestamps are text
    mouse_times = str(
        mouse_times
    )


    # Extract all numerical values from the timestamp string
    timestamp_values = re.findall(
        r'\d+',
        mouse_times
    )


    # Convert timestamp values into integers
    for time_value in timestamp_values:


        try:

            timestamp = int(
                time_value
            )


            # Keep positive timestamp values
            if timestamp > 0:

                mouse_timestamps.append(
                    timestamp
                )


        except ValueError:

            # Ignore values that cannot be converted
            pass


    # --------------------------------------------------------
    # 5. REMOVE TIMESTAMPS WITH AN ABNORMAL DIGIT LENGTH
    # --------------------------------------------------------

    # Run this only when timestamps were found
    if mouse_timestamps:


        # Find the number of digits in every timestamp
        timestamp_lengths = [

            len(
                str(timestamp)
            )

            for timestamp in mouse_timestamps
        ]


        # Find the timestamp length that occurs most often
        most_common_length = max(

            set(
                timestamp_lengths
            ),

            key=timestamp_lengths.count
        )


        # Keep timestamps using the main format
        mouse_timestamps = [

            timestamp

            for timestamp in mouse_timestamps

            if len(
                str(timestamp)
            ) == most_common_length
        ]


    # --------------------------------------------------------
    # 6. CALCULATE MOUSE ACTIVITY DURATION
    # --------------------------------------------------------

    # Start duration as missing
    duration_seconds = np.nan


    # At least two timestamps are required
    if len(mouse_timestamps) >= 2:


        # Find earliest timestamp
        earliest_time = min(
            mouse_timestamps
        )


        # Find latest timestamp
        latest_time = max(
            mouse_timestamps
        )


        # Calculate duration in milliseconds
        duration_milliseconds = (
            latest_time
            -
            earliest_time
        )


        # Convert milliseconds into seconds
        duration_seconds = (
            duration_milliseconds
            /
            1000
        )


        # Replace zero or negative durations
        # with a missing value
        if duration_seconds <= 0:

            duration_seconds = np.nan


    # --------------------------------------------------------
    # 7. CALCULATE AVERAGE MOUSE SPEED
    # --------------------------------------------------------

    # Start average speed as missing
    avg_speed = np.nan


    # Calculate speed only when duration is valid
    if (
        pd.notna(duration_seconds)
        and duration_seconds > 0
    ):


        # Average mouse speed =
        # total mouse distance / duration
        avg_speed = (
            total_distance
            /
            duration_seconds
        )


    # --------------------------------------------------------
    # 8. RETURN THE CALCULATED FEATURES
    # --------------------------------------------------------

    return (

        mouse_events,

        duration_seconds,

        total_distance,

        avg_speed
    )
```

#STEP 9 - Process Phase 1 mouse sessions

```python
# ============================================================
# STEP 9 : PROCESS PHASE 1 MOUSE SESSIONS
# ============================================================


# Create an empty list to store the processed Phase 1 sessions
phase1_session_rows = []


# Search for all Phase 1 mouse_movements.json files
phase1_mouse_files = glob.glob(

    os.path.join(

        # Start from the Phase 1 folder
        phase1_folder,

        # Enter the data folder
        "data",

        # Enter the mouse_movements folder
        "mouse_movements",

        # Match either mouse comparison group folder
        "*",

        # Match every individual session folder
        "*",

        # Select the mouse movement JSON file
        "mouse_movements.json"
    )
)


# Display how many Phase 1 mouse files were found
print(
    "Phase 1 mouse files found:",
    len(phase1_mouse_files)
)


# Create a mapping to make Phase 1 labels
# match the Phase 2 label format
phase1_label_mapping = {

    "human": "Human",

    "moderate_bot": "Moderate Bot",

    "advanced_bot": "Advanced Bot"
}


# Go through every Phase 1 mouse file one at a time
for mouse_file in phase1_mouse_files:


    # Open the current Phase 1 mouse JSON file
    with open(
        mouse_file,
        "r",
        encoding="utf-8"
    ) as file:


        # Read the complete JSON file
        mouse_session_data = json.load(
            file
        )


    # --------------------------------------------------------
    # 1. GET SESSION ID
    # --------------------------------------------------------

    session_id = mouse_session_data.get(
        "session_id"
    )


    # Skip the file if the session ID is missing
    if session_id is None:

        print(
            "Missing session ID in:",
            mouse_file
        )

        continue


    # --------------------------------------------------------
    # 2. FIND SESSION LABEL
    # --------------------------------------------------------

    session_label = phase1_session_labels.get(
        session_id
    )


    # Check whether a label was found
    if session_label is None:

        print(
            "No label found:",
            session_id
        )

        continue


    # Convert Phase 1 labels into the same
    # format used for Phase 2
    session_label = phase1_label_mapping.get(
        session_label,
        session_label
    )


    # --------------------------------------------------------
    # 3. GET RAW MOUSE BEHAVIOUR
    # --------------------------------------------------------

    # Phase 1 stores the complete mouse movement behaviour
    # in the total_behaviour field
    mouse_behaviour = mouse_session_data.get(
        "total_behaviour",
        ""
    )


    # --------------------------------------------------------
    # 4. GET MOUSE TIMESTAMPS
    # --------------------------------------------------------

    mouse_times = mouse_session_data.get(
        "mousemove_times",
        ""
    )


    # --------------------------------------------------------
    # 5. CALCULATE MOUSE FEATURES
    # --------------------------------------------------------

    (
        mouse_events,
        duration_seconds,
        total_distance,
        avg_speed

    ) = calculate_mouse_features(

        mouse_behaviour,
        mouse_times
    )


    # --------------------------------------------------------
    # 6. CREATE ONE SESSION ROW
    # --------------------------------------------------------

    session_row = {

        "session_id": session_id,

        "phase": "Phase 1",

        "label": session_label,

        "mouse_events": mouse_events,

        "duration_seconds": duration_seconds,

        "total_distance": total_distance,

        "avg_speed": avg_speed
    }


    # --------------------------------------------------------
    # 7. ADD SESSION TO PHASE 1 DATA
    # --------------------------------------------------------

    phase1_session_rows.append(
        session_row
    )


# ============================================================
# DISPLAY PHASE 1 RESULTS
# ============================================================


# Display the total number of Phase 1 sessions processed
print(
    "\nPhase 1 sessions processed:",
    len(phase1_session_rows)
)


# Display the Phase 1 class counts
print(
    "\nPhase 1 class counts:"
)


print(
    pd.Series(
        [
            row["label"]
            for row in phase1_session_rows
        ]
    ).value_counts()
)
```

#STEP 10 - Process Phase 2 mouse sessions

```python
# ============================================================
# STEP 10 : PROCESS PHASE 2 MOUSE SESSIONS
# ============================================================


# Create an empty list to store the processed Phase 2 sessions
phase2_session_rows = []


# Create a list containing the three Phase 2 mouse files
# Each file path is paired with its correct class label
phase2_mouse_files = [


    # Human mouse data
    (
        os.path.join(
            phase2_folder,
            "data",
            "mouse_movements",
            "humans",
            "mouse_movements_humans.json"
        ),

        "Human"
    ),


    # Moderate Bot mouse data
    (
        os.path.join(
            phase2_folder,
            "data",
            "mouse_movements",
            "bots",
            "mouse_movements_moderate_bots.json"
        ),

        "Moderate Bot"
    ),


    # Advanced Bot mouse data
    (
        os.path.join(
            phase2_folder,
            "data",
            "mouse_movements",
            "bots",
            "mouse_movements_advanced_bots.json"
        ),

        "Advanced Bot"
    )
]


# Go through each Phase 2 mouse file
# together with its assigned class label
for mouse_file, session_label in phase2_mouse_files:


    # Display which file is currently being processed
    print(
        "\nProcessing:",
        mouse_file
    )


    # Open the current Phase 2 JSON file
    with open(
        mouse_file,
        "r",
        encoding="utf-8"
    ) as file:


        # Phase 2 files contain multiple JSON records
        # Read them one line at a time
        for line in file:


            # Skip empty lines
            if not line.strip():

                continue


            # Convert the current JSON line
            # into a Python dictionary
            mouse_session_data = json.loads(
                line
            )


            # --------------------------------------------------------
            # 1. GET SESSION ID
            # --------------------------------------------------------

            session_id = mouse_session_data.get(
                "session_id"
            )


            # Skip the record if session ID is missing
            if session_id is None:

                print(
                    "Missing session ID in:",
                    mouse_file
                )

                continue


            # --------------------------------------------------------
            # 2. GET RAW MOUSE BEHAVIOUR
            # --------------------------------------------------------

            # Phase 2 stores the mouse movement behaviour
            # in mousemove_total_behaviour
            mouse_behaviour = mouse_session_data.get(
                "mousemove_total_behaviour",
                ""
            )


            # --------------------------------------------------------
            # 3. GET MOUSE TIMESTAMPS
            # --------------------------------------------------------

            mouse_times = mouse_session_data.get(
                "mousemove_times",
                ""
            )


            # --------------------------------------------------------
            # 4. CALCULATE MOUSE FEATURES
            # --------------------------------------------------------

            (
                mouse_events,
                duration_seconds,
                total_distance,
                avg_speed

            ) = calculate_mouse_features(

                mouse_behaviour,
                mouse_times
            )


            # --------------------------------------------------------
            # 5. CREATE ONE SESSION ROW
            # --------------------------------------------------------

            session_row = {

                # Unique session identifier
                "session_id": session_id,

                # Dataset phase
                "phase": "Phase 2",

                # Human, Moderate Bot or Advanced Bot
                "label": session_label,

                # Number of mouse movement events
                "mouse_events": mouse_events,

                # Mouse activity duration in seconds
                "duration_seconds": duration_seconds,

                # Total mouse movement distance
                "total_distance": total_distance,

                # Average mouse movement speed
                "avg_speed": avg_speed
            }


            # --------------------------------------------------------
            # 6. ADD SESSION TO PHASE 2 DATA
            # --------------------------------------------------------

            phase2_session_rows.append(
                session_row
            )


# ============================================================
# DISPLAY PHASE 2 RESULTS
# ============================================================


# Display the total number of Phase 2 sessions processed
print(
    "\nPhase 2 sessions processed:",
    len(phase2_session_rows)
)


# Display the Phase 2 class counts
print(
    "\nPhase 2 class counts:"
)

print(
    pd.Series(
        [
            row["label"]
            for row in phase2_session_rows
        ]
    ).value_counts()
)
```

#STEP 16 - Remove confirmed duplicate sessions
