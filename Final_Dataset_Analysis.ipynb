##5.2 Outlier Analysis
###Step 22: Calculate Simple + Semantic outliers using the IQR method

````python
# ============================================================
# Step 22: Calculate Simple + Semantic outliers using the IQR method
# ============================================================


# Create an empty dictionary to store the outlier count
# for each selected Simple + Semantic feature
simple_semantic_outliers = {}


# Go through each Simple + Semantic feature selected in Step 19
for feature in simple_semantic_visual_features:


    # Calculate the first quartile (25th percentile)
    q1 = (
        simple_semantic_dataframe[
            feature
        ]
        .quantile(
            0.25
        )
    )


    # Calculate the third quartile (75th percentile)
    q3 = (
        simple_semantic_dataframe[
            feature
        ]
        .quantile(
            0.75
        )
    )


    # Calculate the interquartile range
    # IQR is the difference between the third and first quartiles
    iqr = (
        q3 - q1
    )


    # Calculate the lower boundary used to identify outliers
    lower_limit = (
        q1 - (1.5 * iqr)
    )


    # Calculate the upper boundary used to identify outliers
    upper_limit = (
        q3 + (1.5 * iqr)
    )


    # Check which values are below the lower limit
    # or above the upper limit
    outlier_values = (
        (
            simple_semantic_dataframe[
                feature
            ] < lower_limit
        )
        |
        (
            simple_semantic_dataframe[
                feature
            ] > upper_limit
        )
    )


    # Count how many values were identified as outliers
    outlier_count = (
        outlier_values.sum()
    )


    # Store the outlier count using the feature name
    simple_semantic_outliers[
        feature
    ] = outlier_count


# Convert the outlier results into a Pandas Series
# to make the output easier to read
simple_semantic_outlier_results = pd.Series(
    simple_semantic_outliers
)


# Display a heading for the outlier results
print(
    "Simple + Semantic outlier counts:"
)


# Display the number of outliers found for each selected feature
display(
    simple_semantic_outlier_results
)
````

###Step 23: Calculate M4D Mouse outliers using the IQR method

````python
# ============================================================
# Step 23: Calculate M4D Mouse outliers using the IQR method
# ============================================================


# Create an empty dictionary to store the outlier count
# for each selected M4D Mouse feature
m4d_outliers = {}


# Go through each M4D Mouse feature selected in Step 19
for feature in m4d_visual_features:


    # Calculate the first quartile (25th percentile)
    q1 = (
        m4d_mouse_dataframe[
            feature
        ]
        .quantile(
            0.25
        )
    )


    # Calculate the third quartile (75th percentile)
    q3 = (
        m4d_mouse_dataframe[
            feature
        ]
        .quantile(
            0.75
        )
    )


    # Calculate the interquartile range
    # IQR is the difference between the third and first quartiles
    iqr = (
        q3 - q1
    )


    # Calculate the lower boundary used to identify outliers
    lower_limit = (
        q1 - (1.5 * iqr)
    )


    # Calculate the upper boundary used to identify outliers
    upper_limit = (
        q3 + (1.5 * iqr)
    )


    # Check which values are below the lower limit
    # or above the upper limit
    outlier_values = (
        (
            m4d_mouse_dataframe[
                feature
            ] < lower_limit
        )
        |
        (
            m4d_mouse_dataframe[
                feature
            ] > upper_limit
        )
    )


    # Count how many values were identified as outliers
    outlier_count = (
        outlier_values.sum()
    )


    # Store the outlier count using the feature name
    m4d_outliers[
        feature
    ] = outlier_count


# Convert the outlier results into a Pandas Series
# to make the output easier to read
m4d_outlier_results = pd.Series(
    m4d_outliers
)


# Display a heading for the M4D outlier results
print(
    "M4D Mouse outlier counts:"
)


# Display the number of outliers found for each selected feature
display(
    m4d_outlier_results
)
````

##7.2 Human Closeness Analysis
#### Step 30: Calculate exploratory behavioural distance from Human

````python
# ================================================================
# Step 30: Calculate exploratory behavioural distance from Human
# ================================================================

# Calculate the median of each selected M4D behavioural feature for Human, Moderate Bot and Advanced Bot sessions
m4d_group_medians = (
    m4d_mouse_dataframe

    # Group the sessions using their Human or Bot class
    .groupby(
        "label"
    )

    # Select the four M4D behavioural features
    [
        m4d_visual_features
    ]

    # Calculate the median value of every selected feature
    .median()

    # Keep the classes in a clear order
    .reindex(
        [
            "Human",
            "Moderate Bot",
            "Advanced Bot"
        ]
    )
)


# Store the Human median values
# so Human behaviour can be used as the reference
human_medians = (
    m4d_group_medians.loc[
        "Human"
    ]
)


# Select only the Human sessions from M4D
human_m4d_data = (
    m4d_mouse_dataframe[
        m4d_mouse_dataframe[
            "label"
        ] == "Human"
    ]
)


# Calculate the first quartile of every selected Human feature
human_q1 = (
    human_m4d_data[
        m4d_visual_features
    ]

    # Calculate the 25th percentile
    .quantile(
        0.25
    )
)


# Calculate the third quartile of every selected Human feature
human_q3 = (
    human_m4d_data[
        m4d_visual_features
    ]

    # Calculate the 75th percentile
    .quantile(
        0.75
    )
)


# Calculate the Human interquartile range
human_iqr = (
    human_q3
    - human_q1
)


# Replace an IQR of zero with a missing value to prevent division by zero
human_iqr = (
    human_iqr

    # Replace zero values with NumPy missing values
    .replace(
        0,
        np.nan
    )
)


# Create an empty DataFrame to store the Human-distance results
m4d_human_closeness = pd.DataFrame()


# Go through Moderate Bot and Advanced Bot separately
for bot_class in [
    "Moderate Bot",
    "Advanced Bot"
]:

    # Calculate the absolute difference between the current Bot median and Human median
    bot_human_difference = (
        m4d_group_medians.loc[
            bot_class
        ]
        - human_medians
    ).abs()


    # Divide the difference by the Human IQR
    # so features with different measurement scales can be compared more fairly
    normalised_distance = (
        bot_human_difference
        / human_iqr
    )


    # Store the calculated distance values for the current Bot class
    m4d_human_closeness.loc[
        bot_class,
        m4d_visual_features
    ] = normalised_distance


# Calculate one exploratory overall closeness score using the average distance across the four features
m4d_human_closeness[
    "Overall Closeness Score"
] = (
    m4d_human_closeness[
        m4d_visual_features
    ]

    # Calculate the average distance across the four features
    .mean(
        axis=1
    )
)


# Find which Bot class has the smaller distance from Human for every individual feature
closest_bot_per_feature = (
    m4d_human_closeness[
        m4d_visual_features
    ]

    # Find the Bot class with the minimum value
    .idxmin(
        axis=0
    )
)


# Find which Bot class has the smallest overall distance from Human
closest_m4d_bot = (
    m4d_human_closeness[
        "Overall Closeness Score"
    ]

    # Find the class with the lowest overall score
    .idxmin()
)


# Display a heading for the distance results
print(
    "Normalised distance from Human:"
)


# Display the feature-level and overall distance values
display(
    m4d_human_closeness.round(
        3
    )
)


# Display a heading for the feature-level result
print(
    "\nBot class closest to Human for each feature:"
)


# Display which Bot is closer for each individual feature
display(
    closest_bot_per_feature
)


# Display a heading for the overall result
print(
    "\nBot class with the lowest overall exploratory distance from Human:"
)

# Display the Bot class closest to Human overall
print(
    closest_m4d_bot
)


# Explain how the calculated score should be interpreted
print(
    "\nLower closeness score = behaviour closer to Human "
    "across the four selected M4D features."
)


# Create a new figure for the overall closeness result
plt.figure(
    figsize=(7, 5)
)


# Create a bar chart using the overall closeness scores
m4d_human_closeness[
    "Overall Closeness Score"
].plot(
    kind="bar",
    color="darkorange"
)


# Add a title explaining what the graph shows
plt.title(
    "Exploratory Behavioural Distance from Human"
)


# Label the horizontal axis
plt.xlabel(
    "Bot Class"
)


# Label the vertical axis
plt.ylabel(
    "Normalised Distance from Human"
)


# Keep the Bot class names horizontal
plt.xticks(
    rotation=0
)


# Adjust the graph layout
plt.tight_layout()


# Display the closeness bar chart
plt.show()
````
